```python
file_path = "/Volumes/healthcare_readmission/readmission/raw_data/diabetic_data.csv"

df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(file_path)
)

display(df)
```

```python
print("Number of rows:", df.count())
print("Number of columns:", len(df.columns))
```

```python
df.printSchema()
```

```python
display(
    df.groupBy("readmitted")
      .count()
      .orderBy("count", ascending=False)
)
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "healthcare_readmission.readmission.bronze_diabetic_data"
    )
```

```python
bronze_df = spark.table(
    "healthcare_readmission.readmission.bronze_diabetic_data"
)

print("Rows:", bronze_df.count())
print("Columns:", len(bronze_df.columns))

display(bronze_df)
```
